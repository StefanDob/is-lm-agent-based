```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

df = pd.read_csv("islm_daily_output.csv")
df = df.iloc[6000:]

data = df['unmetDemandRatio'].dropna()
xmax = data.max()
bins = np.linspace(0, xmax, 200)

# --- Full view ---
plt.figure(figsize=(8,5))
counts, bins, patches = plt.hist(data, bins=bins, density=True, histtype='step', linewidth=1.2, color='black')

# Convert counts to percentage
counts_percentage = counts / counts.sum() * 100
plt.cla()  # clear current axes
plt.hist(bins[:-1], bins=bins, weights=counts_percentage, histtype='step', linewidth=1.2, color='black')

plt.xlabel("Anteil ungedeckter Nachfrage")
plt.ylabel("Relative Häufigkeit (%)")
plt.title("Verteilung der ungedeckten Nachfrage")
plt.show()

# --- Zoomed-in view ---
# --- Zoomed-in view ---
plt.figure(figsize=(8,5))
counts, bins, patches = plt.hist(data, bins=bins, density=True, histtype='step', linewidth=1.2, color='black')
counts_percentage = counts / counts.sum() * 100
plt.cla()
plt.hist(bins[:-1], bins=bins, weights=counts_percentage, histtype='step', linewidth=1.2, color='black')

plt.xlim(-0.0002, 0.002)  # slightly left of zero
plt.xlabel("Anteil ungedeckter Nachfrage")
plt.ylabel("Relative Häufigkeit (%)")
plt.title("Detailansicht: ungedeckte Nachfrage")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# CSV-Datei laden (Pfad bei Bedarf anpassen)
df = pd.read_csv("islm_semianual_output.csv")

# Sicherstellen, dass keine Leerzeichen in den Spaltennamen enthalten sind
df.columns = df.columns.str.strip()

# ---- 1. Phillips-Kurve (Streudiagramm) ----
plt.figure(figsize=(8, 6))
plt.scatter(df["unemploymentAverage"], df["DeltaPrice"], alpha=0.7, edgecolor="k")

plt.xlabel("Arbeitslosenquote (Halbjahresdurchschnitt)")
plt.ylabel("Preisänderung (Δ Preis, halbjährlich)")
plt.title("Phillips-Kurve (Halbjahresdaten)")
plt.grid(True)
plt.show()
```
