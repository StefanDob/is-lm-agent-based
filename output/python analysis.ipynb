```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV
df = pd.read_csv("islm_daily_output.csv")

# Exclude first 6000 values
df = df.iloc[6000:]

# Convert to percent if needed
data = df[' unmetDemandRatio '] * 100  # Remove *100 if already in percent

# Plot PDF as histogram
plt.figure(figsize=(8, 4))
plt.hist(data, bins=100, density=True, histtype='step', color='black')

plt.xlabel("Unsatisfied demand (in %)")
plt.ylabel("Probability Density Function")
plt.tight_layout()
plt.show()
```

Beveridge Curve:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Excel-Datei einlesen (Tabellenblatt "Monthly Data")
df = pd.read_excel("islm_output.xlsx", sheet_name="Monthly Data")

# Angenommen deine Spalten heißen 'vacancies' und 'unemployed'
vac = df["Offene Stellen"]
unemp = df["Arbeitslose"]

# Reproduzierbare Zufallszahlen (damit das Jitter nicht jedes Mal anders aussieht)
np.random.seed(42)

# Jitter hinzufügen: Uniform zwischen -0.5 und 0.5
vac_jitter = vac + np.random.uniform(-0.5, 0.5, size=len(vac))
unemp_jitter = unemp + np.random.uniform(-0.5, 0.5, size=len(unemp))

# Scatterplot mit Jitter
plt.figure(figsize=(8,6))
plt.scatter(vac_jitter, unemp_jitter, alpha=0.5, s=20)
plt.xlabel("Vakanz")
plt.ylabel("Arbeitslose")
plt.title("Beveridge-Kurve mit Jitter")
plt.show()


```
