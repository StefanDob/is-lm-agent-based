```python
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("islm_daily_output.csv")
df = df.iloc[6000:]

data = df['unmetDemandRatio'].dropna()
xmax = data.max()
bins = np.linspace(0, xmax, 200)

# --- Full view ---
plt.figure(figsize=(8,5))
counts, bins, patches = plt.hist(data, bins=bins, density=True, histtype='step', linewidth=1.2, color='black')

# Convert counts to percentage
counts_percentage = counts / counts.sum() * 100
plt.cla()  # clear current axes
plt.hist(bins[:-1], bins=bins, weights=counts_percentage, histtype='step', linewidth=1.2, color='black')

plt.xlabel("Unsatisfied demand")
plt.ylabel("Percentage (%)")
plt.title("Full Distribution")
plt.show()

# --- Zoomed-in view ---
# --- Zoomed-in view ---
plt.figure(figsize=(8,5))
counts, bins, patches = plt.hist(data, bins=bins, density=True, histtype='step', linewidth=1.2, color='black')
counts_percentage = counts / counts.sum() * 100
plt.cla()
plt.hist(bins[:-1], bins=bins, weights=counts_percentage, histtype='step', linewidth=1.2, color='black')

plt.xlim(-0.0002, 0.002)  # slightly left of zero
plt.xlabel("Unsatisfied demand")
plt.ylabel("Probability Density Function (%)")
plt.title("Zoomed-in Distribution")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV (adjust path if needed)
df = pd.read_csv("islm_semianual_output.csv")

# Ensure no stray whitespace in column names
df.columns = df.columns.str.strip()

# ---- 1. Phillips Curve Scatter ----
plt.figure(figsize=(8, 6))
plt.scatter(df["unemploymentAverage"], df["DeltaPrice"], alpha=0.7, edgecolor="k")
plt.xlabel("Unemployment Rate (quarterly average)")
plt.ylabel("Price Inflation (Δ Price, quarterly)")
plt.title("Phillips Curve (Quarterly Data)")
plt.grid(True)
plt.show()
```
