```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

df = pd.read_csv("islm_daily_output.csv")
df = df.iloc[6000:]

data = df['unmetDemandRatio']

# Use density=True to get PDF
counts, bins, _ = plt.hist(data, bins=100, density=True, histtype='step', color='black')

# Convert PDF to percentage for y-axis
counts_percent = counts * 100
bin_width = bins[1] - bins[0]

print("Area under histogram =", np.sum(counts * bin_width))

# Plot with percentages
plt.clf()  # Clear previous plot
plt.step(bins[:-1], counts_percent, where='mid', color='black')
plt.xlabel("Unsatisfied demand")
plt.ylabel("Probability Density (%)")
plt.tight_layout()
plt.show()


```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load CSV (adjust path if needed)
df = pd.read_csv("islm_semianual_output.csv")

# Ensure no stray whitespace in column names
df.columns = df.columns.str.strip()

# ---- 1. Phillips Curve Scatter ----
plt.figure(figsize=(8, 6))
plt.scatter(df["unemploymentAverage"], df["DeltaPrice"], alpha=0.7, edgecolor="k")
plt.xlabel("Unemployment Rate (quarterly average)")
plt.ylabel("Price Inflation (Δ Price, quarterly)")
plt.title("Phillips Curve (Quarterly Data)")
plt.grid(True)
plt.show()
```
